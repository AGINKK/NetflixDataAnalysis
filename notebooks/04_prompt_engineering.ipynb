```python
import pandas as pd
from pathlib import Path

project_dir = Path.cwd().parent

df = pd.read_csv(
    project_dir / "data" / "processed" / "netflix_clean.csv"
)

print("Dataset loaded:", df.shape)
```

```python
basic_prompt = """
Analyze the Netflix dataset.
"""

improved_prompt = """
Act as an experienced data analyst.

Context:
I am analyzing a Netflix Movies and TV Shows dataset
containing title metadata.

Task:
Identify data-quality issues and suggest appropriate checks.

Constraints:
- Do not invent statistics.
- Do not delete records automatically.
- Explain missing values and duplicates separately.
- Use beginner-friendly Python and Pandas.
- Preserve the original dataset.

Expected output:
1. Data-quality checks
2. Suggested cleaning actions
3. Python code
4. Verification steps
"""

print("BASIC PROMPT:\n", basic_prompt)
print("\nIMPROVED PROMPT:\n", improved_prompt)
```

```python
quality_report = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (
        df.isna().sum() / len(df) * 100
    ).round(2)
})

print("Dataset shape:", df.shape)
print("Exact duplicate rows:", df.duplicated().sum())
print("\nData-quality report:")

display(quality_report)
```

```python
type_counts = df["type"].value_counts(dropna=False)

print(type_counts)
print("\nTotal records:", len(df))
print("Count total:", type_counts.sum())

assert type_counts.sum() == len(df)

print("Verification passed: content-type counts reconcile.")
```
