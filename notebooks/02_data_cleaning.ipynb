```python
import pandas as pd
from pathlib import Path

project_dir = Path.cwd().parent
file_path = project_dir / "data" / "raw" / "netflix_titles.csv"

df = pd.read_csv(file_path)

print("DATASET SIZE:", df.shape)

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nEXACT DUPLICATE ROWS")
print(df.duplicated().sum())

print("\nDUPLICATE SHOW IDs")
print(df["show_id"].duplicated().sum())

print("\nCONTENT TYPES")
print(df["type"].value_counts(dropna=False))

print("\nRATING VALUES")
print(df["rating"].value_counts(dropna=False).head(15))

print("\nDURATION VALUES")
print(df["duration"].value_counts(dropna=False).head(15))
```

```python
clean_df = df.copy()

text_columns = clean_df.select_dtypes(include="object").columns

for col in text_columns:
    clean_df[col] = clean_df[col].map(
        lambda x: x.strip() if isinstance(x, str) else x
    )

clean_df["date_added"] = pd.to_datetime(
    clean_df["date_added"],
    format="mixed",
    errors="coerce"
)

clean_df = clean_df.drop_duplicates()

output_dir = project_dir / "data" / "processed"
output_dir.mkdir(parents=True, exist_ok=True)

clean_df.to_csv(
    output_dir / "netflix_clean.csv",
    index=False
)

print("Original rows:", len(df))
print("Cleaned rows:", len(clean_df))
print("Rows removed:", len(df) - len(clean_df))
print("Cleaned file saved successfully")
```
